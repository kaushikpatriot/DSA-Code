##String Matching Algorithms

### Brute Force - Forward scan - O(nm)

```python
def stringmatch(t,p):
  poslist = []
  for i in range (len(t) - len(p) + 1):
    j = 0
    matched = True
    while j < len(p) and matched:
      if(t[i+j] != p[j]):
        matched = False
      j += 1
    if matched:
      poslist.append(i)
  return poslist

t = 'straw plus berry is strawberry'
p = 'strawberry'
print (stringmatch(t,p))
```

### Brute Force - Backward scan - O(nm)

```python
def stringmatchrev(t,p):
  poslist = []
  for i in range (len(t) - len(p) + 1):
    j = len(p) - 1
    matched = True
    while j >= 0 and matched:
      if(t[i+j] != p[j]):
        matched = False
      j -= 1
    if matched:
      poslist.append(i)
  return poslist

t = 'which finally halts.  at that point'
p = 'at that point'
print (stringmatchrev(t,p))
```

### Boyer Moore Algorithm - O(nm)

Backward scan with skipping technique

```python
def boyermoore(t,p):
  # Dictionary storing the index of character in the pattern (p)
  m = len(p)
  match_count = 0
  last = {p[i]:i for i in range(m)}
  # Initialize poslist and i , as skipping of characters will be based on matches
  poslist, i = [], 0
  skipL = []
  while i <= (len(t) - len(p)):
    j , matched = len(p) - 1, True
    skipL.append(i)
    while j >= 0 and matched:
      match_count += 1
      if (t[i+j] != p[j]):
        matched = False
      j = j - 1
    if matched:
      poslist.append(i)
      i = i + 1
    else:
      j = j + 1
      if (t[i+j] in last.keys()):
        i = i + max(j - last[t[i+j]],1) # Max used to ensure that a char in t is not matched with a later char in p, thus pushing p back
      else:
        i = i + j + 1
  #print (match_count, skipL)
  return (poslist)

t = 'abaaabaaaaacdabab'
p = 'aa'
#t = 'straw plus berry is strawberry'
#p = 'strawberry'
#t = 'lcetcxedt dfashoxdwkevfbiztvrwh xqhjtxntfplurlpkrbpvgehojnkagvqla'
#p = 'ojnka'
print (boyermoore(t,p))
```

###Rabin Karp Algorithm - Simple
O(nm) - worst case - usually O(m)

```python
def rabinkarp(t,p):
  poslist = []
  numt, nump = 0, 0
  for i in range (len(p)):
    numt = 10*numt + int(t[i])
    nump = 10*nump + int(p[i])

  if (numt == nump):
    poslist.append(0)

  for i in range (1, len(t)-len(p)+1): #Already zero-th position compared, hence starting at 1
    numt = numt - int(t[i-1]) * (10**(len(p)-1))
    numt = 10*numt + int(t[i+len(p)-1])
    if (numt==nump):
      poslist.append(i)
  return poslist

t = '64564335677829937824'
p = '782'
print (rabinkarp(t,p))
```

### String Tries
**Trie structure**
1. There will be root that consists of a dictionary
2. There will a unique letter in each node
3. Traversing through the trie will give the word searched for.
4. The end of a valid word has a $ node at the end of it

**Implementation**
1. Trie is implemented as Class with the following functions
2. Init, add word, query

```python
class Trie:
  def __init__(self,S=[]):
    self.root = {}
    for s in S:
      self.add(s)

  def add(self,s):
    curr = self.root
    if len(s) < 1:
      return
    s = s + '$'
    for c in s:
      if c not in curr.keys():
        curr[c] = {}
      curr = curr[c]

  def __str__(self):
    return str(self.root)

  def query(self,s):
    curr = self.root
    if len(s) < 1:
      return None
    for c in s:
      if (c not in curr.keys()):
        return False
      curr = curr[c]
    if '$' in curr.keys():
      return True
    else:
      return None

t = Trie()
t.add('best')
t.add('bet')
t.add('better')
t.add('bell')
print(t)
print(t.query('better'))
print(t.query('bello'))
```

```python
class SuffixTrie():
  def __init__(self,s):
    self.root  = {}
    s = s + '$'
    for i in range (len(s)):
      curr = self.root
      for c in s[i:]:
        if (c not in curr.keys()):
          curr[c] = {}
        curr = curr[c]

  def followPath(self,s):
    curr = self.root
    for c in s:
      if (c not in curr.keys()):
        return None
      curr = curr[c]
    return curr

  def hasSuffix(self,s):
    node = self.followPath(s)
    return (node is not None and '$' in node.keys())

  def __str__(self):
    return str(self.root)

t = SuffixTrie('ababab')
print (t)
print(t.hasSuffix('aba'))
print(t.hasSuffix('abab'))
```

### Grpa - 3 - Make Palindrome

```python
def MakePalindrome(s):
  prefix = ''
  i = 0
  j = len(s)
  while i < j:
    if s[i] == s[j-1]:
      i += 1
      j -= 1
    else:
      prefix += s[j-1]
      j -= 1
  if prefix == '':
    prefix = None
  return prefix

s = 'malayalam'
print(MakePalindrome(s))
```

##PPA-2 - Possible Prefix matches

```python
def PrefixMatch(s):
  n = len(s)
  fail = [0 for i in range(n)]
  j, k = 1, 0
  while j < n:
    if s[j] == s[k]:
      fail[j] = k+1
      j, k = j+1, k+1
    elif k > 0:
      k = fail[k-1]
    else:
      j = j+1
  prefix = []
  for i in fail:
    l = s[:i]
    if l != '' and l not in prefix:
      prefix.append(l)
  return prefix

s = 'abababbbbbabababbbaaaabbb'
print (PrefixMatch(s))
```
