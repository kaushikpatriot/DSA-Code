## AVL Trees

```python
class AVLTree():
  def __init__(self,initval=None):
    self.value = initval
    if(self.value):
      self.left = AVLTree()
      self.right = AVLTree()
      self.height = 1
    else:
      self.left = None
      self.right = None
      self.height = 0
    return

  def isempty(self):
    return (self.value == None)

  def isleaf(self):
    return (self.value != None and self.left.isempty() and self.right.isempty())

  def find(self,v):
    if self.isempty():
      return (False)
    if self.value == v:
      return (True)
    if v < self.value:
      return (self.left.find(v))
    if v > self.value:
      return (self.right.find(v))

  def minval(self):
    if self.left.isempty():
      return (self.value)
    else:
      return (self.left.minval())

  def maxval(self):
    if self.right.isempty():
      return (self.value)
    else:
      return (self.right.maxval())

  def insert(self,v):
    if self.isempty():
      self.value = v
      self.left = AVLTree()
      self.right = AVLTree()
      self.height = 1
    if self.value == v:
      return
    if v < self.value:
      self.left.insert(v)
      self.rebalance()
      self.height = 1 + max(self.left.height,self.right.height)
      return
    if v > self.value:
      self.right.insert(v)
      self.rebalance()
      self.height = 1 + max(self.left.height,self.right.height)
      return

  def delete(self,v):
    if self.isempty():
      return
    if v < self.value:
      self.left.delete(v)
      self.height = 1 + max(self.left.height,self.right.height)
      return
    if v > self.value:
      self.right.delete(v)
      self.height = 1 + max(self.left.height,self.right.height)
      return
    if v == self.value:
      if self.isleaf():
        self.makeempty()
        self.height = 0
      elif self.left.isempty():
        self.copyright()
        self.height -= 1
      elif self.right.isempty():
        self.copyleft()
        self.height -= 1
      else:
        self.value = self.left.maxval()
        self.left.delete(self.left.maxval())
      return

  def makeempty(self):
    self.value = None
    self.left = None
    self.right = None
    return

  def copyleft(self):
    self.value = self.left.value
    self.right = self.left.right
    self.left = self.left.left
    return

  def copyright(self):
    self.value = self.right.value
    self.left = self.right.left
    self.right = self.right.right
    return

  def leftrotate(self):
    v = self.value
    vr = self.right.value
    tl = self.left
    trl = self.right.left
    trr = self.right.right
    newtree = AVLTree(v)
    newtree.left = tl
    newtree.right = trl
    self.value = vr
    self.right = trr
    self.left = newtree
    return

  def rightrotate(self):
    v = self.value
    vl = self.left.value
    tr = self.right
    tll = self.left.left
    tlr = self.left.right
    newtree = AVLTree(v)
    newtree.left = tlr
    newtree.right = tr
    self.right = newtree
    self.value = vl
    self.left = tll
    return

  def rebalance(self):
    slope = self.left.height - self.right.height
    #case 1: slope = 0 Tree is balanced
    if slope in (-1,0,1):
      return
    #case 2: slope > 1 Tree is imbalanced on the left side
    elif slope > 1:
      if self.left.left.height > self.left.right.height:
        self.rightrotate()
      if self.left.left.height < self.left.right.height:
        self.left.leftrotate()
        self.rightrotate()
      self.updateheight()
    #case 3: slope < -1 Tree is imbalanced on the right side
    else:
      if self.right.right.height > self.right.left.height:
        self.leftrotate()
      if self.right.right.height < self.right.left.height:
        self.right.rightrotate()
        self.leftrotate()
      self.updateheight()

  def updateheight(self):
    if self.isempty():
      return
    else:
      self.left.updateheight()
      self.right.updateheight()
      self.height = 1 + max(self.left.height,self.right.height)

  def inorder(self):
    if self.isempty():
      return ([])
    else:
      return (self.left.inorder() + [self.value] + self.right.inorder())

  def preorder(self):
    if self.isempty():
      return ([])
    else:
      return ([self.value] + self.left.preorder() + self.right.preorder())

  def postorder(self):
    if self.isempty():
      return ([])
    else:
      return (self.left.postorder() + self.right.postorder() + [self.value])

  def __str__(self):
    return (str(self.inorder()))


A = AVLTree()
nodes = [1,2,3,4,5,6,7]
for i in nodes:
    A.insert(i)

print(A.inorder())
print(A.preorder())
print(A.postorder())
```

##Interval Scheduling

```python
def intervalschedule(L):
  L1 = []
  L2 = []
  L3 = []
  count = 0
  lastfinish = 0
  for i,j,k in L:
    L1.append((k,j,i))
  L1.sort()
  for k,j,i in L1:
    L2.append((i,j,k))

  for i,j,k in L2:
    if (j > lastfinish):
      L3.append(i)
      count+=1
      lastfinish = k

  return count

L = [(0, 1, 2),(1, 1, 3),(2, 1, 5),(3, 3, 4),(4, 4, 5),(5, 5, 8),(6, 7, 9),(7, 10, 13),(8, 11, 12)]
print(intervalschedule(L))
```

##Minimizing Lateness

```python
from operator import itemgetter

def minimize_lateness(jobs):
  schedule = []
  max_lateness = 0
  lastfinish = 0
  sorted_jobs = sorted(jobs, key=itemgetter(2))

  for j,t,d in sorted_jobs:
    start = lastfinish
    lastfinish = start + t
    schedule.append((j,start, lastfinish))
    if (max(lastfinish - d,0) > max_lateness): #Negative delays not allowed
      max_lateness = lastfinish - d

  return max_lateness, schedule

jobs = [(1, 3, 6), (2, 2, 9), (3, 1, 8), (4, 4, 9), (5, 3, 14), (6, 2, 15)]
print (minimize_lateness(jobs))
```

##Huffman Coding

```python
class Node:
    def __init__(self,frequency,symbol = None,left = None,right = None):
        self.frequency = frequency
        self.symbol = symbol
        self.left = left
        self.right = right

  # Creating the Huffman coding Tree
  # Create a list of tuples consisting of each character and its frequency - Frequency list
  # Create a list of tuples that consist of each tuple above and the corresponding Node in the Tree - Nodes List
  # Iteratively combine each of the tuples and create a node for the same along with links to its children nodes
    # Sort the nodes list in ascending
    # Left node = Nodelist [0][1]
    # Right node = Nodelist[1][1]
    # Create a new node as [Left node frequence + Right node frequency, left node symbol + right node symbol, left node, right node]
    # Delete the first two items from the nodes list
  # Codifying
    # Start with the root node - Last node from the previous step
    # For each unique character
    # process until the character does not match witht he symbol in the node
    # if the character is in the left node symbol, add 0 to the code and move left
    # if the character is in the right node symbol, add 1 to the code and move right
    # Add the final code against the character in the dictionary

def Huffman(s):
  # Create a unique set of character
  uniq_list = set(s)
  freq_list = []
  nodes_list = []
  for ch in uniq_list:
    freq_list.append((s.count(ch),ch))
  #freq_list.sort()
  for tup in sorted(freq_list):
    nodes_list.append((tup,Node(tup[0],tup[1])))
  while len(nodes_list) > 1:
    nodes_list.sort()
    L = nodes_list[0][1]
    R = nodes_list[1][1]
    new_node = Node(L.frequency+R.frequency,L.symbol+R.symbol,L,R)
    nodes_list.pop(0)
    nodes_list.pop(0)
    nodes_list.append(((L.frequency+R.frequency,L.symbol+R.symbol), new_node))

  huffcode = {}
  for ch in uniq_list:
    temp = nodes_list[0][1]
    code = ''
    while ch != temp.symbol:
      if (ch in temp.left.symbol):
        code += '0'
        temp = temp.left
      else:
        code += '1'
        temp = temp.right
    huffcode[ch] = code

  return huffcode

s = 'aaaacccbbdddddddeeefffff'
res = Huffman(s)
#print (res)
for char in sorted(res):
    print(char,res[char])

```

##Grpa - 3 - Scheduling problem##

```python
def no_overlap(L):
  L1 = []
  for (i,j,k) in L:
    L1.append((k,j,i))
  L1.sort()
  curr_day = 0
  cust_list = []
  for (l,m,n) in L1:
    if curr_day == 0 :
      curr_day = l
      cust_list.append(n)
    else:
      if m > curr_day:
        cust_list.append(n)
        curr_day = l
  return cust_list

#L = [(0, 1, 2), (1, 1, 3), (2, 1, 5), (3, 3, 4), (4, 4, 5), (5, 5, 8), (6, 7, 9), (7, 10, 13), (8, 11, 12)]

L = [(0, 2, 4),(1, 1, 6),(2, 4, 10),(3, 9, 10),(4, 1, 2),(5, 7, 8),(6, 4, 5),(7, 6, 10),(8, 9, 15),(9, 8, 13)]
print(no_overlap(L))
```
