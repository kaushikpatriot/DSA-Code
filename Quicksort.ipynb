##Quick Sort##

```python
def quickSort(L,l,r):
# Identify the first element as the pivot
# Classify from left to right as Lower and Upper
# Place the pivot in the middle
# Recursively Quick sort the Lower and Upper separately
  if (r-l <= 1):
    return
  else:
    (pivot,lower,upper) = (L[l],l+1,l+1)
    for i in range(l+1,r):
      if (L[i] > pivot): # classified as Upper
        upper += 1
      elif (L[i] <= pivot): # classified as Lower
        (L[i],L[lower]) = (L[lower],L[i])
        (lower,upper) = (lower+1,upper+1)
  (L[l],L[lower-1]) = (L[lower-1],L[l]) #placing pivot between lower and upper
  lower -= 1
  quickSort(L,l,lower)
  quickSort(L,lower+1,upper)
  return L
```

```python
L = [1,3,5,0,2,4,17,2,-5,6,4,3]
n = len(L)
print (quickSort(L,0,n))
```

##Linked List##

```python
class Node():
  # Node initialised with given value
  def __init__(self,v=None):
    self.value = v
    self.next = None

  # Check if the list is empty
  def isEmpty(self):
    if (self.value == None):
      return True
    return False

  # Append an item
  def append(self,v):
    if (self.value == None):
      self.value = v
    else:
      p = Node(v)
      self.next = p

  # Insert an item at the head
  def insert(self,v):
    # if list is empty
    if (self.isEmpty()):
      self.append(v)
    else:
      n = Node(self.value)
      # Singleton list
      if (self.next == None):
        self.value = v
        self.next = n
      else:
        self.value = v
        n.next = self.next
        self.next = n

  def delete(self,v):
    if (self.isEmpty()):
      return

    if (self.value == v):
      self.value = None
      if self.next != None:
        self.value = self.next.value
        self.next = self.next.next
      return
    else:
      if self.next != None:
        self.next.delete(v)
        if self.next.value == None:
          self.next = None
    return



  # Display List
  def display(self):
    print (id(self),self.value,self.next)
    if (self.next != None):
      self.next.display()



```
